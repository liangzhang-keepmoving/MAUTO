```python

from env_simplified import SimplifiedMultiUAVEnvironment
```

```python
env = SimplifiedMultiUAVEnvironment(
        num_uavs=3, 
        num_users=5,
        trajectory_file="user_trajectories.json"
    )
```

```python
print(f"总目标任务大小: {env.total_target_task_size} MB")
print(f"预期episode步数: {env.target_episode_steps}")
```

```python
from DDPG import DDPGAgent
agent = DDPGAgent(n_uavs=3, n_users=5)
```

```python
state = env.reset()
allocation, offloading, motion = agent.select_action(state,hard=False)
print(allocation)
print(offloading)
print(motion)
```

```python
observations, total_rewards, done, info, uav_delays = env.step([allocation, offloading, motion])
print(uav_delays)
```
