```python
import torch
import numpy as np
from DDPG import DDPGAgent
from env_simplified import SimplifiedMultiUAVEnvironment
env = SimplifiedMultiUAVEnvironment(
        num_uavs=3, 
        num_users=5,
        trajectory_file="user_trajectories.json"
    )
agent = DDPGAgent(n_uavs=3, n_users=5)
```

```python
state = env.reset()
allocation = torch.tensor([[0.3514, 0.3364, 0.3331, 0.3211, 0.3264],
                           [0.3261, 0.3368, 0.3375, 0.3405, 0.3282],
                           [0.3226, 0.3268, 0.3294, 0.3384, 0.3454]])

offloading = torch.tensor([[0.1794, 0.1607, 0.1705, 0.1550, 0.1571],
                           [0.1676, 0.1737, 0.1751, 0.1664, 0.1625],
                           [0.1653, 0.1668, 0.1682, 0.1630, 0.1751]])

motion = torch.tensor([[0.4419, 0.1020],
                       [0.4419, 0.0000],
                       [0.4764, 0.0176]])
```

```python
observations, total_rewards, done, info, uav_delays = env.step([allocation, offloading, motion])
print([allocation, offloading, motion])
print(uav_delays)
```

```python
state = env.reset()
actions = {}
for uav_id in range(env.num_uavs):
    actions[f'uav_{uav_id}'] = {
        'user_competition_probs': allocation[uav_id].cpu().numpy(),
        'offloading_ratios': offloading[uav_id].cpu().numpy(),
        'movement_direction': motion[uav_id, 1].item() * np.pi,  # 转换为弧度
        'movement_distance': motion[uav_id, 0].item() * agent.max_distance
    }
observations, total_rewards, done, info, uav_delays = env.step(actions)
print(uav_delays)
```
